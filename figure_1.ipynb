```python
import numpy as np
import matplotlib as mpl
import matplotlib.pyplot as plt

from color import base_colors, darken_color, saturate_color, vivid_color


mpl.rcParams["figure.dpi"] = 600


np.random.seed(11)

n = 100
x = np.linspace(-3, 3, n)

y_a = x + np.random.normal(0, 0.6, n)
y_b = 0.9 * x + 0.3 * np.tanh(x) + np.random.normal(0, 0.7, n)


fig, ax = plt.subplots(figsize=(0.7, 0.7), dpi=600)

c_main = darken_color(base_colors[1],0.9)
c_aux = vivid_color(base_colors[3],0.9)

ax.scatter(y_a, y_b, s=3, alpha=1, color=c_main, edgecolors="none")
ax.scatter(y_a[::8], y_b[::8], s=3, alpha=1, color=c_aux, edgecolors="none")

ax.plot([-4, 4], [-4, 4],
        linewidth=0.6,
        linestyle="--",
        color=base_colors[0],
        alpha=0.6)

ax.set_xticks([])
ax.set_yticks([])
ax.set_xlabel("")
ax.set_ylabel("")
ax.set_frame_on(False)

plt.tight_layout(pad=0.00)
plt.show()
```

```python
import numpy as np
import matplotlib as mpl
import matplotlib.pyplot as plt

from color import base_colors, darken_color, saturate_color, vivid_color

mpl.rcParams["figure.dpi"] = 600

np.random.seed(11)

def make_clusters(centers, n_per=80, scale=0.35):
    pts_all = []
    for cx, cy in centers:
        pts = np.random.normal(loc=[cx, cy], scale=scale, size=(n_per, 2))

        # mild TSNE-like warp
        x, y = pts[:, 0], pts[:, 1]
        pts[:, 0] = x + 0.25 * np.sin(y)
        pts[:, 1] = y + 0.25 * np.cos(x)

        pts_all.append(pts)

    return np.vstack(pts_all)


data_1 = make_clusters([(-2, -1), (0, 1.2), (2, -0.5)])

fig1, ax1 = plt.subplots(figsize=(0.7, 0.7), dpi=600)

c1 = darken_color(base_colors[1], 0.9)

ax1.scatter(data_1[:, 0], data_1[:, 1],
            s=3, color=c1, edgecolors="none", alpha=1)

ax1.set_xticks([])
ax1.set_yticks([])
ax1.set_frame_on(False)
ax1.margins(0)

plt.tight_layout(pad=0.0)
plt.show()


data_2 = make_clusters([(-1.5, 1.5), (1.8, 1.0), (0.2, -1.8)])

fig2, ax2 = plt.subplots(figsize=(0.7, 0.7), dpi=600)

c2 = vivid_color(base_colors[3], 0.9)

ax2.scatter(data_2[:, 0], data_2[:, 1],
            s=3, color=c2, edgecolors="none", alpha=1)

ax2.set_xticks([])
ax2.set_yticks([])
ax2.set_frame_on(False)
ax2.margins(0)

plt.tight_layout(pad=0.0)
plt.show()
```
