```python

def display_plots(pre_title, data, labels):

    # plot the views
    plt.figure()
    fig, ax = plt.subplots(1,2, figsize=(14,5))
    dot_size=10
    ax[0].scatter(data[0][:, 0], data[0][:, 1],c=labels,s=dot_size)
    ax[0].set_title(pre_title + ' View 1')
    ax[0].axes.get_xaxis().set_visible(False)
    ax[0].axes.get_yaxis().set_visible(False)

    ax[1].scatter(data[1][:, 0], data[1][:, 1],c=labels,s=dot_size)
    ax[1].set_title(pre_title + ' View 2')
    ax[1].axes.get_xaxis().set_visible(False)
    ax[1].axes.get_yaxis().set_visible(False)

    plt.show()
```

```python
import numpy as np

np.random.seed(42)  # For reproducibility

num_blocks = 4
block_size = 100
dataset_length = len(train_dataset)

random_indices_chunked = []

for _ in range(num_blocks):
    # Choose a random start index such that the block fits in the dataset
    start = np.random.randint(0, dataset_length - block_size + 1)
    block = list(range(start, start + block_size))
    random_indices_chunked.extend(block)

print(random_indices_chunked)
```

```python
m_data = [
    np.stack([grad.half().numpy() for grad in estimators[0].get_gradient(train_dataset_name, train_dataset_split, random_indices_chunked)])
    for j in range(len(estimators))
]


epsilon = 1e-5  
m_data_affinity = []
for X in m_data:
    affinity = rbf_kernel(X)          
    affinity += epsilon * np.eye(X.shape[0])  
    m_data_affinity.append(affinity)
```

```python

m_spectral1 = MultiviewSpectralClustering(n_clusters=4,
            random_state=RANDOM_SEED, n_init=100)
m_clusters1 = m_spectral1.fit_predict(m_data_affinity)
```
